```python
for run in df['r_id'].unique():
    plt.figure(figsize=(8, 6))
    sns.lineplot(
        data=df[df['r_id'] == run],
        x='time', y='vcd',
        errorbar=None
    )
    plt.title(f'Run {run}')
    plt.xlabel('Time [day]')
    plt.ylabel('VCD [10⁶ cells/mL]')
    plt.grid(True)
```

```python
for clone in df['clone_id'].unique():
    plt.figure(figsize=(8, 6))
    sns.lineplot(
        data=df[df['clone_id'] == clone],
        x='time', y='vcd',
        hue='r_id',
        style ='group_volume',
        errorbar=None
    )
    plt.title(f'Clone {clone}')
    plt.xlabel('Time [day]')
    plt.ylabel('VCD [10⁶ cells/mL]')
    plt.grid(True)
```

```python
numeric_df = df.select_dtypes(include=['number']).drop(columns=['p_id', 'r_id', 'clone_id', 'group_volume'])
corr = numeric_df.corr()
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, fmt='.2f')
```
