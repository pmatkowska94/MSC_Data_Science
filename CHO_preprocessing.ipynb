```python
train_df = pd.read_csv('train_df.csv')
test_df = pd.read_csv('test_df.csv')
val_df = pd.read_csv('val_df.csv')
```

**Missing Values**

kla > 50 % will be dropped from dataset

np and impeller_no are variables used for engineering features but irrelevant in the dataset, will be dropped

critical information missing: time, vcd, volume will be dropped

vvm is important and will be imputed

```python
train_df.drop(columns=['kla', 'impeller_d', 'np', 'do', 'impeller_no', 'agitation'], inplace=True)
test_df.drop(columns=['kla', 'impeller_d', 'np', 'do', 'impeller_no', 'agitation'], inplace=True)
val_df.drop(columns=['kla', 'impeller_d', 'np', 'do', 'impeller_no', 'agitation'], inplace=True)
```

```python
train_df.info()
```

```python
train_df = train_df.dropna(subset=['vcd', 'time'])
test_df = test_df.dropna(subset=['vcd', 'time'])
val_df = val_df.dropna(subset=['vcd', 'time'])
```

```python
train_df.info()
```

```python
train_df = train_df.dropna(subset=['volume'])
test_df = test_df.dropna(subset=['volume'])
val_df = val_df.dropna(subset=['volume'])
```

```python
train_df.info()
```

```python
print(train_df[train_df['tip_speed'].isna() | train_df['reynolds'].isna()])
```

```python
train_df = train_df.dropna(subset=['tip_speed'])
```

```python
train_df.info()
```

```python
test_df.info()
```

```python
val_df.info()
```

All critical missing values were excluded, including time, VCD, and volume. Redundant variables such as impeller number, impeller power number, agitation, and impeller diameter were also removed. kLa was excluded due to over 50% missing data. DO was treated as a setpoint rather than a process variable.

```python
train_df = train_df.drop(columns=['vessel'])
test_df = test_df.drop(columns=['vessel'])
val_df = val_df.drop(columns=['vessel'])
```

**Missing VVM**

```python
train_df.groupby('p_id')['vvm'].mean()
```

```python
train_df.groupby('p_id')['vvm'].median()
```

```python
train_df.groupby('group_volume')['vvm'].median()
```

```python
train_df['vvm'] = train_df.groupby('p_id')['vvm'].transform(lambda x: x.fillna(x.median()))
train_df['vvm'] = train_df.groupby('group_volume')['vvm'].transform(lambda x: x.fillna(x.median()))
train_df['vvm'] = train_df['vvm'].fillna(0.02)
```

```python
train_df.groupby('group_volume')['vvm'].median()
```

Missing vvm values were imputed using a biologically informed approach. They were first filled with the median per process (p_id), then with the median per reactor scale (group_volume), and finally replaced with a typical CHO value of 0.02 VVM.

```python
test_df['vvm'] = test_df.groupby('p_id')['vvm'].transform(lambda x: x.fillna(x.median()))
test_df['vvm'] = test_df.groupby('group_volume')['vvm'].transform(lambda x: x.fillna(x.median()))
test_df['vvm'] = test_df['vvm'].fillna(0.02)
```

```python
val_df['vvm'] = val_df.groupby('p_id')['vvm'].transform(lambda x: x.fillna(x.median()))
val_df['vvm'] = val_df.groupby('group_volume')['vvm'].transform(lambda x: x.fillna(x.median()))
val_df['vvm'] = val_df['vvm'].fillna(0.02)
```

```python
test_df.groupby('group_volume')['vvm'].median()
```

```python
val_df.groupby('group_volume')['vvm'].median()
```

**Missing aspect ratio**

```python
train_df.groupby('p_id')['aspect_r'].median()
```

```python
train_df.groupby('group_volume')['aspect_r'].median()
```

```python
train_df['aspect_r'] = train_df.groupby('p_id')['aspect_r'].transform(lambda x: x.fillna(x.median()))
train_df['aspect_r'] = train_df.groupby('group_volume')['aspect_r'].transform(lambda x: x.fillna(x.median()))
train_df['aspect_r'] = train_df['aspect_r'].fillna(2.0)
```

```python
test_df['aspect_r'] = test_df.groupby('p_id')['aspect_r'].transform(lambda x: x.fillna(x.median()))
test_df['aspect_r'] = test_df.groupby('group_volume')['aspect_r'].transform(lambda x: x.fillna(x.median()))
test_df['aspect_r'] = test_df['aspect_r'].fillna(2.0)
```

```python
val_df['aspect_r'] = val_df.groupby('p_id')['aspect_r'].transform(lambda x: x.fillna(x.median()))
val_df['aspect_r'] = val_df.groupby('group_volume')['aspect_r'].transform(lambda x: x.fillna(x.median()))
val_df['aspect_r'] = val_df['aspect_r'].fillna(2.0)
```

Missing aspect_r values were imputed using same as vvm approach. They were first filled with the median per process (p_id), then with the median per reactor scale (group_volume), and finally replaced with a default value of 2.0, typical for stirred-tank bioreactors.
