```python
stages = ["Initial LSTM", "BiLSTM", "Transfer Learning"]
MAE = [3.32, 2.04, 3.01]
RMSE = [4.32, 3.01, 4.05]
RMSE_pct = [17.3, 12.1, 36.1]

x = np.arange(len(stages))

plt.figure(figsize=(8, 5))

plt.bar(x, MAE, label='MAE', zorder=3)
plt.bar(x, np.array(RMSE) - np.array(MAE), bottom=MAE, label='RMSE', zorder=3)
for i, val in enumerate(RMSE):
    plt.text(x[i], val + 0.15, f"RMSE {RMSE_pct[i]}%",
             ha='center', va='bottom', fontsize=9, color='black',  weight='bold')
plt.title("Pilot Study Model Performance", fontsize=14, weight='bold')
plt.ylabel("Error (×10^6 cells/mL)", fontsize=12)
plt.xticks(x, stages, fontsize=10)
plt.legend(frameon=False)
plt.grid(axis='y', linestyle='--', alpha=0.4, zorder=0)
plt.tight_layout()
plt.ylim(0, 6)
```
